```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------
# Parámetros
# ------------------------
#del modelo
K1 = 1
sigma1 = 1

# de las distribuciones
lambda_exp = K1   # parámetro de la exponencial
alpha = K1 / sigma1        # parámetro de forma de Lomax
theta = 1 / sigma1        # parámetro de escala de Lomax

# ------------------------
# Dominio
# ------------------------
x = np.linspace(0, 15, 500)

# ------------------------
# Distribuciones
# ------------------------

# Exponencial: f(x) = lambda * exp(-lambda x)
exp_dist = lambda_exp * np.exp(-lambda_exp * x)

# Lomax: f(x) = (alpha/theta) * (1 + x/theta)^(-(alpha+1))
lomax_dist = (alpha / theta) * (1 + x / theta) ** (-(alpha + 1))

# ------------------------
# Gráfica
# ------------------------
plt.figure()
plt.plot(x, exp_dist,linestyle='--' , label='Exponential',linewidth=3)
plt.plot(x, lomax_dist, label='Lomax', linewidth=3)

plt.xlim(0,7)
plt.ylim(0, 1)

plt.xticks(fontsize=15)  # Eje X
plt.yticks(fontsize=15)  # Eje Y

plt.xlabel('$T$', fontsize=20)
plt.legend(prop={'size':17})
plt.grid()

plt.show()
```
