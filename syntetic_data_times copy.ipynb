Times in the indiferent and the curious state

```python
import scipy.stats as stats
import numpy as np

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import lomax
from scipy.stats import kstest

# Parameters
n = 200     # size of data

K1 = 1
sigma1 = 0.25
alpha1 = K1 / sigma1   # form
lambda1 = 1 / sigma1 # scale


K2 = 2
sigma2 = 0.25
alpha2 = K2 / sigma2   # form
lambda2 = 1 / sigma2 # scale

# obtained synthetic data
T_plus = stats.lomax.rvs(c=alpha1, scale=lambda1, size=n)

T_minus = stats.lomax.rvs(c=alpha2, scale=lambda2, size=n)
```

```python
############################################################
# AJUSTE LOMAX
############################################################

# loc=0 porque los tiempos comienzan en cero

shape_A, loc_A, scale_A = lomax.fit(T_plus, floc=0)


K_plus= shape_A/scale_A
sigma_plus= 1/scale_A

print(shape_A/scale_A, 1/scale_A)
#print(shape_I/scale_I, 1/scale_I)

#error
print((K_plus-K1)*100/K1, (sigma_plus-sigma1)*100/sigma1) 
```

```python
#determing values of parameters without extrinsic fluctuations
media_T_plus = sum(T_plus) / n
K_plus_alt= 1/media_T_plus

print(K_plus_alt)

#error
print((K_plus_alt-K1)*100/K1)
```

```python
############################################################
# AJUSTE LOMAX
############################################################

# loc=0 porque los tiempos comienzan en cero

shape_I, loc_I, scale_I = lomax.fit(T_minus, floc=0)


K_minus= shape_I/scale_I
sigma_minus= 1/scale_I

#print(shape_A/scale_A, 1/scale_A)
print(shape_I/scale_I, 1/scale_I)

#error
print((K_minus-K2)*100/K2, (sigma_minus-sigma2)*100/sigma2) 
```

```python
#determing values of parameters without extrinsic fluctuations
media_T_minus = sum(T_minus) / n
K_minus_alt= 1/media_T_minus

print(K_minus_alt)

#error
print((K_minus_alt-K2)*100/K2)
```
