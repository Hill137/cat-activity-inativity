```python
import numpy as np
import matplotlib.pyplot as plt

def telegraph_process(k_I, k_A, T, S0=0):
    
    t = 0
    state = S0
    
    times = [0]
    states = [state]
    
    while t < T:
        
        if state == 0:
            rate = k_I
            next_state = 1
        else:
            rate = k_A
            next_state = 0
        
        # tiempo de espera exponencial
        tau = np.random.exponential(1/rate)
        
        t += tau
        
        if t > T:
            break
        
        state = next_state
        
        times.append(t)
        states.append(state)
    
    return np.array(times), np.array(states)


# Parámetros
k_I = 1.0
k_A = 2
T = 50

times, states = telegraph_process(k_I, k_A, T)

# Graficar
plt.step(times, states, where="post", linewidth=3)
plt.xlabel("Time", fontsize=20)
plt.ylabel("State", fontsize=20)
plt.yticks([0,1])
plt.xlim(0,50)
plt.ylim(-0.01,1.01)
#plt.title("Simulación del proceso de telégrafo")
plt.show()
```
